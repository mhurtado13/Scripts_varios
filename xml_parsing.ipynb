```python
import xml.etree.ElementTree as ET
```

```python
tree = ET.parse('Physicell_NLCs-CLL/output/initial.xml')
```

## Parsing XML

```python
root = tree.getroot()
root.tag
root.attrib
```

```python
for child in root:
    print(child.tag, child.attrib)
```

```python
root[1][0].tag #first index corresponds to first child, second index to nested child
root[1][0].attrib
root[1][0][1].tag
```

```python
for neighbor in root.iter('diffusion_coefficient'):
    print(neighbor.text)
```

```python
for country in root.findall('country'):
    rank = country.find('rank').text
    name = country.get('name')
    print(name, rank)
```

## Modifying an XML file

#### Modifying elements and attributes

```python
for rank in root.iter('rank'):
    new_rank = int(rank.text) + 1
    rank.text = str(new_rank)
    rank.set('updated', 'yes')

tree.write('output.xml')
```

#### Removing elements

```python
for country in root.findall('country'):
    rank = int(country.find('rank').text)
    if rank > 50:
        root.remove(country)

tree.write('output.xml')
```
