```python
from scipy.optimize import fsolve

# Function to calculate the future value (FV) of an investment with initial savings
def calculate_future_value(P_monthly, r_annual, n_years, initial_savings=0):
    r_monthly = r_annual / 12
    n_months = n_years * 12
    FV = initial_savings * (1 + r_monthly)**n_months + P_monthly * (((1 + r_monthly)**n_months - 1) / r_monthly)
    return FV

# Function to solve for the number of years (n) to reach the target FV with initial savings
def calculate_n(P_monthly, r_annual, FV_target, initial_savings=0):
    r_monthly = r_annual / 12
    
    # Define the equation to solve
    def equation(n):
        n_months = 12 * n
        return initial_savings * (1 + r_monthly)**n_months + P_monthly * (((1 + r_monthly)**n_months - 1) / r_monthly) - FV_target
    
    # Initial guess for n
    n_guess = 30
    
    # Use fsolve to find the value of n that solves the equation
    n_solution = fsolve(equation, n_guess)
    return n_solution[0]
```

Calculate future value after a specific number of years

```python
# Variables
P_monthly = 2000  # Monthly contribution in EUR
r_annual = 0.07  # Annual interest rate (7%)
n_years_for_FV = 20  # Example: Calculate FV after 30 years
FV_after_30_years = calculate_future_value(P_monthly, r_annual, n_years_for_FV)
print(f"Future Value after {n_years_for_FV} years: €{FV_after_30_years:.2f}")
```

Calculate the number of years required to reach the target FV

```python
# Variables
P_monthly = 200  # Monthly contribution in EUR
r_annual = 0.07  # Annual interest rate (7%)
FV_target = 1000000  # Target future value in EUR

n_years_to_target = calculate_n(P_monthly, r_annual, FV_target)
print(f"Years to reach €{FV_target}: {n_years_to_target:.2f} years")
```
